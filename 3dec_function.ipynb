**args and kwargs**


- args means we can provide multiple values **without argument name**   ex. func('abcd',12)
   - output will come in **tupple format** like ('abcd',12)
- keyword arguments means we can provide multiple values **with argument name** ex. func(name='abcd',age=12)
   - output will come in **dictionary format** like  {'name':'abcd', 'age' :12}

```python
def arg(*args):
    print(args)
```

```python
arg(10,20)
```

```python
def arg(*args):
    a = args[0]
    b= args[1]
    return a,b
num= arg(10,29)
```

```python
num
```

```python
def arg2(**kwargs):
    print(kwargs)

arg2(a=10,b=20)
```
