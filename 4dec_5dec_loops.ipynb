**Loops**

- loops are used to execute a block of code multiple times
- 3 paramerters
    -  start point
    -  increment/decrement
    -  condition to stop
- for loop and while loop
- if know the condition to stop then for loop
- if we dont know when to stop then while loop

**Case-3 range(start,stop,step)**

- start value = start
- if step is +ve
    - it is postivie direction
    - increment by **step value**
    - last value= stop -1
- if step is -ve
   - it is -ve direction
    - decrement by **step value**
    - last value= stop + 1

```python
for i in range(1,20,2):
    print(i ,end =' ')

#start = 1
#step = + 2
#stop =stop -1 =20-1 =19
```

```python
for i in range(20,1, -2):
    print(i ,end =' ')

#start = 20
#direction = -2  -ve direction
#stop = 1+1 =2 
```

```python
for i in range(20,1,2):
    print(i ,end =' ')

#start =20
#increment=2
#stop = 1-1 = 0    #not possible
```

```python
for i in range(-2,-20,2):
    print(i ,end =' ')

#start =-2
#step +2
#stop = -20-1 =-21   #not possible
```

**Continue pass break**


```python
for i in range(5):             #step wil be last-1 i.e till 4
    if i == 2:
        print('great')
    print(i)
# i=0   0==2   f   print(0)
# i=1   1==2   f   print 1
# i=2   2==2  t   print ('great') print (2)
# i=3   3==2  f   print 3
#i=4    4==2  f   print 4

    
```

```python
for i in range(5):             #step wil be last-1 i.e till 4
    if i == 2:
        print('great')
        continue
    print(i)
```

```python
for i in range(5):             #step wil be last-1 i.e till 4
    if i == 2:
        continue
        print('great')
    print(i)
```

```python
for i in range(5):             #step wil be last-1 i.e till 4
    if i == 2:
        print('great')
        continue
        print('super')
    print(i)
print('ok')
```

```python
for i in range(5):             #step wil be last-1 i.e till 4
    if i == 2:
        pass
        print('great')
    print(i)
```

- **continue** will skip the next line directly goes to next iteration
- **pass** just allow the loop execution
- **break** will stop the execution
